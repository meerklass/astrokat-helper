The code below calculates the angular distance in ra,dec after 400 seconds for a starting position given in ra,dec. Drift scan assumed (constant az,el)

```python
# source 3C273,  radec= 12:29:6.6996 02:03:8.5968
#ra=360/24*(12+29/60+6.6996/60/60)*u.deg
#dec=(2+3/60+8.5968/60/60)*u.deg

# name=3C353, radec= 17:20:28.16 -00:58:46.62
ra=360/24*(17+20/60+28.16/60/60)*u.deg
dec=(0-58/60-46.62/60/60)*u.deg

#ra=180*u.deg
#dec=-70*u.deg

print(ra,dec)
```

```python
source_radec = SkyCoord(ra=ra, dec=dec, frame='icrs')
```

```python
start_time=Time(val='2024-05-17 20:43:12', scale='utc')
```

```python
duration=400*u.second
#duration=200*u.second/np.cos(dec)

print(duration)
```

```python
cazel=source_radec.transform_to(AltAz(location=EarthLocation.of_site('MEERKAT'), obstime=start_time+duration/2))
```

```python
alt=cazel.alt
az=cazel.az
print(alt,az)
print(alt*u.degree,az*u.degree)
```

```python
cazel1 = SkyCoord(alt=alt, az=az, frame='altaz', location=EarthLocation.of_site('MEERKAT'), obstime=start_time)
```

```python
cradec1=cazel1.icrs
```

```python
print(cradec1.ra,cradec1.dec)
print(cradec1.ra*u.degree,cradec1.dec*u.degree)
```

```python
cazel2 = SkyCoord(alt=alt, az=az, frame='altaz', location=EarthLocation.of_site('MEERKAT'), obstime=start_time+duration)
```

```python
cradec2=cazel2.icrs
```

```python
print(cradec2.ra,cradec2.dec)
print(cradec2.ra*u.degree,cradec2.dec*u.degree)
```

```python
print(cradec2.separation(cradec1).deg)
print(cradec2.separation(source_radec).deg)
print(cradec1.separation(source_radec).deg)
```
